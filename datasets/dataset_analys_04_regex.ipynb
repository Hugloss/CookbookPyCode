# Regex dataset with pandas

```python
import re
import pandas as pd
```

```python
# path to folder
path_dir = '../datasets/dataset/full_text_dataset.csv'
```

### Check regex one by one

```python
text1 = "Intyg från revisor med korrekt datum för fusionsplan bifogas. "
patterns = r"intyg\s+från\s+r[eo]v[il1][sl]or" # Intyg från revisor / rovisor

pattern = re.compile(patterns, re.IGNORECASE | re.VERBOSE)
matches = re.findall(pattern, text1)
matches
```

```python
text2 = "Det här är ett revisorsintyg som styrker att en oberoende granskning av företagets räkenskaper har genomförts enligt god revisionssed. "
patterns = r"r[eo]visors?int[vy]g" # Revisorsintyg, intvg

pattern = re.compile(patterns, re.IGNORECASE | re.VERBOSE)
matches = re.findall(pattern, text2)
matches
```

```python
text3 = "Revisorns yttrande bekräftade att företagets årsredovisning gav en rättvisande bild av dess ekonomiska ställning. "
patterns = r"r[eo]v[il1][sl]ors?n?s?\s*y*t*tr?a*n?d?e" # Revisorsyttrande + OCR variants

pattern = re.compile(patterns, re.IGNORECASE | re.VERBOSE)
matches = re.findall(pattern, text3)
matches
```

### Combine regexs

```python
text = text1 + text2 + text3
patterns = r"""
    intyg\s+från\s+r[eo]v[il1][sl]or               # Intyg från revisor / rovisor
    | r[eo]visors?int[vy]g                         # Revisorsintyg, intvg
    | r[eo]v[il1][sl]ors?n?s?\s*y*t*tr?a*n?d?e     # Revisorsyttrande + OCR variants         
"""
# patterns = r"intyg\s+från\s+r[eo]v[il1][sl]or| r[eo]visors?int[vy]g| r[eo]v[il1][sl]ors?n?s?\s*y*t*tr?a*n?d?e"
pattern = re.compile(patterns, re.IGNORECASE | re.VERBOSE)
matches = re.findall(pattern, text)
matches
```

### Load dataset

```python
df = pd.read_csv(path_dir)
```

```python
# df.head()
```

### find regex match in text

```python
len(df)
```

```python
df_selection = df[(df['class'] == 'YTTRANDE') & (df['pagenr'] == 0)].copy()
len(df_selection)
```

```python
# Apply to each row, case-insensitive
df_selection['matches'] = df_selection['text'].apply(lambda x: re.findall(pattern, str(x)))
```

### view data for none match

```python
matches_df = df_selection[df_selection['matches'].apply(lambda x: len(x) > 0)]
len(matches_df)
```

```python
no_matches_df = df_selection[df_selection['matches'].apply(lambda x: len(x) == 0)]
len(no_matches_df)
```
