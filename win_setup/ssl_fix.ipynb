# Refresh your CA bundle

The ca-bundle.pem file is your organization’s custom certificate bundle. On SUSE it is generated on the server under /var/lib/ca-certificates/ and symlinked to /etc/ssl/ca-bundle.pem. When Python or other tools throw an SSL verification error, it usually means this file is outdated or missing your company’s root CA. To resolve the error, export a fresh ca-bundle.pem from the Linux server (for example by running sudo update-ca-certificates or copying the file from /var/lib/ca-certificates) and replace the old bundle on your client machine.

```python
requests.get("https://www.google.com", verify=r"C:\_INSTALL\ca-bundle.pem").raise_for_status()
```
