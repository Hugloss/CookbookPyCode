```python
import re
import pandas as pd
from collections import Counter
```

```python
import stopwordsiso as stopwords

swedish_stopwords = stopwords.stopwords("sv")
# print(swedish_stopwords)
```

```python
text = [
        'Mötet är planerat till den 15/04/2023 klockan 14:00.',
        'Försäljningen ökade med 15 procent under tredje kvartalet 2023.',
        'Avtalet löper ut den 2024-01-31 efter midnatt.',
        'Vi behöver gå igenom de ekonomiska rapporterna för 2022 i detalj.',
        'Kundnöjdheten förbättrades markant i region 3 enligt senaste mätningen.'
    ]
text
```

```python
def preprocess_text(text):
    """
    Preprocess the text by removing dates, numbers, and other unwanted characters.
    
    Args:
        text (str): Input text to preprocess
        
    Returns:
        str: Preprocessed text
    """
    if not isinstance(text, str):
        return ""
    
    # Remove dates (various formats)
    text = re.sub(r'\b\d{1,2}[/-]\d{1,2}[/-]\d{2,4}\b', '', text)
    text = re.sub(r'\b\d{4}[/-]\d{1,2}[/-]\d{1,2}\b', '', text)
    
    # Remove all numbers
    text = re.sub(r'\b\d+\b', '', text)
    
    # Remove extra whitespace
    text = re.sub(r'\s+', ' ', text).strip()
    tokens = text.split()
    return tokens
```

```python
# Build Counter-based BoW
bow_counters = [Counter(preprocess_text(sentence)) for sentence in text]

# Show the result
for i, bow in enumerate(bow_counters):
    print(f"Dokument {i+1}:")
    print(bow)
    list_of_words = list(bow.keys())
    print(list_of_words)
    print(text[i])
    print()
```

### Remove swedish stopwords

```python
def preprocess_text_swe(text):
    """
    Preprocess the text by removing dates, numbers, and other unwanted characters.
    
    Args:
        text (str): Input text to preprocess
        
    Returns:
        str: Preprocessed text
    """
    if not isinstance(text, str):
        return []
    
    # Convert to lowercase
    text = text.lower()
    
    # Remove dates (various formats)
    text = re.sub(r'\b\d{1,2}[/-]\d{1,2}[/-]\d{2,4}\b', '', text)
    text = re.sub(r'\b\d{4}[/-]\d{1,2}[/-]\d{1,2}\b', '', text)
    
    # Remove all numbers
    text = re.sub(r'\b\d+\b', '', text)
    
    # Remove punctuation
    text = re.sub(r'[^\w\s]', '', text)
    
    # Remove extra whitespace
    text = re.sub(r'\s+', ' ', text).strip()
    
    # Tokenize and filter stopwords
    tokens = text.split()
    tokens = [word for word in tokens if word not in swedish_stopwords]
    
    return tokens
```

```python
# Build Counter-based BoW
bow_counters_swe = [Counter(preprocess_text_swe(sentence)) for sentence in text]

# Show the result
for i, bow in enumerate(bow_counters_swe):
    print(f"Dokument {i+1}:")
    print(bow)
    list_of_words = list(bow.keys())
    print(list_of_words)
    print(text[i])
    print()
```

### pandas preprocessing

```python
# create dataframe
df = pd.DataFrame({
    'text': text
})

# Apply preprocessing to the text column
df['tokens'] = df['text'].apply(preprocess_text_swe)
df
```
