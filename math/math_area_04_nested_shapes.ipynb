# 🔷 Geometric Setup: Rectangle → Ellipse → Rectangle

We are exploring how three nested shapes relate in terms of area:

1. A **big rectangle**
2. An **ellipse** inscribed inside it
3. A **smaller rectangle** inside that ellipse

---

## 🔹 Step 1: Big Rectangle (Outer Box)

* Dimensions: $2a$ wide and $2b$ high
* Area:

$$
A_{\text{big\_rect}} = 2a \cdot 2b = 4ab
$$

---

## 🔹 Step 2: Ellipse Inside the Rectangle

* The ellipse fits perfectly inside the rectangle and touches all sides
* Its semi-axes are $a$ and $b$
* Area:

$$
A_{\text{ellipse}} = \pi ab
$$

---

## 🔹 Step 3: Small Rectangle Inside the Ellipse

* The largest rectangle that fits inside the ellipse, aligned with its axes
* Area:

$$
A_{\text{small\_rect}} = 2ab
$$

---

### 📝 Why the Max Rectangle Inside an Ellipse Has Area $2ab$

This is a known result from **calculus**:

The general equation of an ellipse is:

$$
\frac{x^2}{a^2} + \frac{y^2}{b^2} = 1
$$

If we want to fit a rectangle inside this ellipse (with sides parallel to the axes), we try to maximize the area of such a rectangle.

* It turns out the **maximum** happens when the corners of the rectangle touch the ellipse at the points:

  $$
  \left(\frac{a}{\sqrt{2}}, \frac{b}{\sqrt{2}}\right)
  $$

So, the side lengths of the rectangle are:

* Width: $2 \cdot \frac{a}{\sqrt{2}} = \frac{2a}{\sqrt{2}}$
* Height: $2 \cdot \frac{b}{\sqrt{2}} = \frac{2b}{\sqrt{2}}$

Area:

$$
A = \frac{2a}{\sqrt{2}} \cdot \frac{2b}{\sqrt{2}} = \frac{4ab}{2} = 2ab
$$

✔️ This confirms our earlier result.

---

## 📊 Area Ratios

| Shape           | Area     | Compared to Previous | Compared to Big Rectangle |
| --------------- | -------- | -------------------- | ------------------------- |
| Big Rectangle   | $4ab$    | —                    | 100%                      |
| Ellipse         | $\pi ab$ | \~78.5%              | \~78.5%                   |
| Small Rectangle | $2ab$    | \~63.7%              | 50%                       |

---

## 📘 Bonus: Why the Inner Rectangle Has Sides $\frac{a}{\sqrt{2}}$ and $\frac{b}{\sqrt{2}}$

To find the **largest rectangle** that fits inside the ellipse:

$$
\frac{x^2}{a^2} + \frac{y^2}{b^2} = 1
$$

We optimize the area of a rectangle with half-width $x$ and half-height $y$ under this constraint.

The solution gives:

$$
x = \frac{a}{\sqrt{2}}, \quad y = \frac{b}{\sqrt{2}}
$$

So the full side lengths of the rectangle are:

$$
\text{Width} = 2 \cdot \frac{a}{\sqrt{2}} = \frac{2a}{\sqrt{2}}, \quad
\text{Height} = 2 \cdot \frac{b}{\sqrt{2}} = \frac{2b}{\sqrt{2}}
$$

✅ Area Check:

$$
\text{Area} = \left(\frac{2a}{\sqrt{2}}\right) \cdot \left(\frac{2b}{\sqrt{2}}\right)
= \frac{4ab}{2} = 2ab
$$

✔️ Which matches our earlier result.

---

```python
def calculate_inscribed_shapes(x, y):
    # Outer rectangle area
    outer_rectangle_area = 4 * x * y

    # Ellipse area (inscribed in rectangle)
    ellipse_area = np.pi * x * y

    # Max area rectangle inscribed in ellipse
    inner_rectangle_area = 2 * x * y

    # Side lengths of the inner rectangle
    x_inner = x / np.sqrt(2)
    y_inner = y / np.sqrt(2)
    inner_width = 2 * x_inner
    inner_height = 2 * y_inner

    # Assert the area matches (allowing a tiny margin for floating-point precision)
    computed_area = inner_width * inner_height
    assert np.isclose(inner_rectangle_area, computed_area, atol=1e-6), \
        "Area does not match the product of side lengths!"

    # Area ratio
    ratio = outer_rectangle_area / inner_rectangle_area

    return (
        outer_rectangle_area,
        ellipse_area,
        inner_rectangle_area,
        ratio,
        inner_width,
        inner_height
    )

def print_inscribed_shapes(results):
    (
        outer_area,
        ellipse_area,
        inner_area,
        ratio,
        inner_width,
        inner_height
    ) = results

    print(f"Outer rectangle area: {outer_area}")
    print(f"Ellipse area (inscribed in rectangle): {ellipse_area:.2f}")
    print(f"Rectangle inscribed in ellipse (max area): {inner_area}")
    print(f"Area ratio (outer rect / inner-inner rect): {ratio:.2f}")
    print(f"Inner rectangle side lengths: {inner_width:.2f} x {inner_height:.2f}")
```

```python
# Example usage:
x = 3
y = 5
results = calculate_inscribed_shapes(x, y)
print_inscribed_shapes(results)
```
