# Batch run OCR

```python
import os
import re
from PIL import Image
import pytesseract
from docx import Document
```

```python
def get_all_image_files(folder_path):
    """
    Retrieve all image files from the given folder.

    :param folder_path: Path to the folder containing image files.
    :return: List of file paths.
    """
    supported_formats = ('.png', '.jpg', '.jpeg', '.bmp', '.tiff')
    image_files = [os.path.join(folder_path, f) for f in os.listdir(folder_path) if f.lower().endswith(supported_formats)]
    return image_files

def extract_text_from_image(image_path):
    """
    Perform OCR on the given image and extract text.

    :param image_path: Path to the image file.
    :return: Extracted text as a string.
    """
    image = Image.open(image_path)
    text = pytesseract.image_to_string(image, lang="swe")

    # Ensure the string is XML-compatible
    text = re.sub(r'[^\x09\x0A\x0D\x20-\x7E\xA0-\uD7FF\uE000-\uFFFD]', '', text)

    return text

def write_text_to_file(text_data, output_text_file):
    """
    Write the extracted text to a text file.

    :param text_data: Text data to write.
    :param output_text_file: Path to the output text file.
    """
    with open(output_text_file, 'w') as text_file:
        text_file.write(text_data)

def write_text_to_word(text_data, output_word_file):
    """
    Write the extracted text to a Word document.

    :param text_data: Text data to write.
    :param output_word_file: Path to the output Word document.
    """
    doc = Document()
    doc.add_paragraph(text_data)
    doc.save(output_word_file)

def process_images_in_folder(folder_path, output_text_file, output_word_file):
    """
    Process all images in a folder: perform OCR and save the results to text and Word files.

    :param folder_path: Path to the folder containing image files.
    :param output_text_file: Path to the output text file.
    :param output_word_file: Path to the output Word document.
    """
    image_files = get_all_image_files(folder_path)
    all_text = ""

    for image_file in image_files:
        print(f"Processing {image_file}...")
        text = extract_text_from_image(image_file)
        all_text += f"\n\n\n{'='*40}\n{image_file}\n{text}"

    write_text_to_file(all_text, output_text_file)
    write_text_to_word(all_text, output_word_file)
```

```python
# folder_path = 'datasets/example'
output_text_file = 'datasets/ocr_output_text.txt'
output_word_file = 'datasets/ocr_output_text.docx'

process_images_in_folder(folder_path, output_text_file, output_word_file)
```
