```python
import base64
import ollama
```

```python
# Load the image
img_path = '../_data/resource/note_child.jpg'
```

```python
base_url="http://localhost:11434/"
```

```python
prompt_default = "What's in this image?"
prompt_custom = "Extract all text, focusing on dates and names."
prompt_system = """Extract text, return tables in HTML, equations in LaTeX. Wrap watermarks/signatures/page numbers in tags, describe images, use ☐ and ☑ for checkboxes."""
```

```python
# select model
model_ocr = "benhaotang/Nanonets-OCR-s"
```

```python
# Initialize Ollama client
client = ollama.Client(host=base_url)
```

```python
response = client.chat(
    model=model_ocr,
    messages=[{
        'role': 'user',
        'content': prompt_default,
        'images': [img_path]
    }]
)

print(response['message']['content'])
```

```python
# base64 encode image
def encode_image(image_path):
    with open(image_path, "rb") as image_file:
        return base64.b64encode(image_file.read()).decode('utf-8')

# Encode your image
encoded_image = encode_image(img_path)
```

```python
response = client.chat(
    model=model_ocr,
    messages=[{
        'role': 'user',
        'content': prompt_default,
        'images': [encoded_image]
    }]
)

print(response['message']['content'])
```
