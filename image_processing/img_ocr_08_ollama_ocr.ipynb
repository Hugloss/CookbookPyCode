```python
from ollama_ocr import OCRProcessor
```

```python
# Load the image
img_path = '../_data/resource/note_child.jpg'
```

```python
base_url="http://localhost:11434/"
```

```python
prompt_default = "What's in this image?"
prompt_custom = "Extract all text, focusing on dates and names."
prompt_system = """Extract text, return tables in HTML, equations in LaTeX. Wrap watermarks/signatures/page numbers in tags, describe images, use ☐ and ☑ for checkboxes."""
```

```python
# select model
model_ocr = "benhaotang/Nanonets-OCR-s"
```

```python
# Initialize OCR processor
ocr = OCRProcessor(model_name=model_ocr, base_url=base_url + "/api/generate")
```

```python
# Process an image
result = ocr.process_image(
    image_path=img_path, # path to your pdf files "path/to/your/file.pdf"
    format_type="markdown",  # Options: markdown, text, json, structured, key_value
    preprocess=False,  # Enable image preprocessing
    custom_prompt=prompt_default, # Optional custom prompt
    language="English" # Specify the language of the text
)
print(result)
```
